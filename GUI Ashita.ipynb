```python
import tkinter as tk
from tkinter import messagebox, ttk
import pandas as pd
import numpy as np
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split
from catboost import CatBoostRegressor
import matplotlib.pyplot as plt
from matplotlib.backends.backend_tkagg import FigureCanvasTkAgg
```

```python
df = pd.read_csv(r'D:\2025 Work\Ashita\Work Dr Ashita Data\Data 70;30\Dr Ashita Data Central composite design (CCD).csv', encoding='ISO-8859-1')
```

```python
X = df.drop(columns=['Compressive Strength (MPa)'])
```

```python
y = df['Compressive Strength (MPa)']
```

```python
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
```

```python
X_train, X_test, y_train, y_test = train_test_split(X_scaled, y, test_size=0.3, random_state=42)
```

```python
model = CatBoostRegressor(iterations=500, learning_rate=0.1, depth=6, random_state=42, verbose=0)
model.fit(X_train, y_train)
```

```python
def predict_Compressive_Strength():
    try:
        inputs = [float(entry.get()) for entry in entries]
        input_scaled = scaler.transform([inputs])
        prediction = model.predict(input_scaled)[0]
        result_var.set(f"{prediction:.2f}")

        if prediction < 5:
            result_entry.config(fg='green')
        elif prediction < 10:
            result_entry.config(fg='orange')
        else:
            result_entry.config(fg='red')
    except ValueError:
        messagebox.showerror("Error", "Please enter valid numbers in all fields")

def clear_inputs():
    for entry in entries:
        entry.delete(0, tk.END)
    result_var.set("")
    result_entry.config(fg='black')
```

```python
root = tk.Tk()
root.title("Compressive Strength")
root.geometry("800x600")
root.configure(bg='#f0f0f0')

style = ttk.Style()
style.configure('TFrame', background='#f0f0f0')
style.configure('TLabel', background='#f0f0f0', font=('Arial', 10))
style.configure('TButton', font=('Arial', 10), padding=5)
style.configure('Header.TLabel', font=('Arial', 12, 'bold'))

# Header
header = ttk.Frame(root)
header.pack(pady=10)
ttk.Label(header, text="Compressive Strength", style='Header.TLabel').pack()

# Input fields
input_frame = ttk.Frame(root)
input_frame.pack(pady=10)

entries = []
for i, label in enumerate(X.columns):
    frame = ttk.Frame(input_frame)
    frame.grid(row=i//2, column=i%2, padx=15, pady=5, sticky="ew")
    ttk.Label(frame, text=label + ":", width=25).pack(side='left')
    entry = tk.Entry(frame, width=15)
    entry.pack(side='left')
    entries.append(entry)

# Buttons
button_frame = ttk.Frame(root)
button_frame.pack(pady=15)

ttk.Button(button_frame, text="Predict Compressive Strength", command=predict_Compressive_Strength).pack(side='left', padx=5)
ttk.Button(button_frame, text="Clear", command=clear_inputs).pack(side='left', padx=5)

# Result display
result_frame = ttk.Frame(root)
result_frame.pack(pady=10)

ttk.Label(result_frame, text="Predicted Compressive Strength:").pack(side='left')
result_var = tk.StringVar()
result_entry = tk.Entry(result_frame, textvariable=result_var, font=('Arial', 12, 'bold'), fg='black', width=15, justify='center')
result_entry.pack(side='left', padx=5)
```

```python
footer = ttk.Frame(root)
footer.pack(pady=10)
ttk.Label(footer, text="CatBoostRegressor Model").pack()

root.mainloop()
```
